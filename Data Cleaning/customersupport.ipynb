```python
import pandas as pd
import re

# Load dataset
df = pd.read_csv("aa_dataset-tickets-multi-lang-5-2-50-version (1).csv")

df = df.drop_duplicates()

df.columns = (
    df.columns
      .str.strip()
      .str.lower()
      .str.replace(" ", "_")
)

text_cols = df.select_dtypes(include="object").columns

def clean_text(text):
    if pd.isna(text):
        return text
    text = text.strip()
    text = re.sub(r"\s+", " ", text)          # remove extra spaces
    text = re.sub(r"[^\w\s.,!?]", "", text)   # remove weird symbols
    return text

for col in text_cols:
    df[col] = df[col].apply(clean_text)

df["priority"] = df["priority"].str.lower()

# Type
df["type"] = df["type"].str.title()

# Queue
df["queue"] = df["queue"].str.title()

# Language
df["language"] = df["language"].str.lower()

df[text_cols] = df[text_cols].fillna("Not Available")

# 6. Remove empty / useless records
df = df[~((df["subject"] == "Not Available") & (df["body"] == "Not Available"))]

# 7. Reset index
df.reset_index(drop=True, inplace=True)

# 8. Save cleaned data
df.to_csv("cleaned_support_data.csv", index=False)

print("Data cleaning completed successfully")
df
```
